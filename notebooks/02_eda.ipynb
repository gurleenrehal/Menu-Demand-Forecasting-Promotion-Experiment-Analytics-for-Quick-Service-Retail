```python
import sys; sys.path.insert(0, '../src')
from data_preprocessing import run as preprocess
from visualization import plot_monthly_revenue, plot_revenue_by
import matplotlib.pyplot as plt
%matplotlib inline

df, _ = preprocess('../data/coffee_shop_sales.csv')
df.head()
```

```python
plot_monthly_revenue(df);
```

Revenue climbs steadily from January ($81.7K) to June ($166.5K) - roughly doubling over the window, consistent with a growing but still-young store footprint (or a seasonal spring/summer effect that a longer history would need to confirm).

```python
plot_revenue_by(df, 'store_location');
```

```python
plot_revenue_by(df, 'product_category');
```

Coffee and Tea alone account for roughly two-thirds of total revenue. Packaged Chocolate, Flavours and Loose Tea are long-tail categories.
