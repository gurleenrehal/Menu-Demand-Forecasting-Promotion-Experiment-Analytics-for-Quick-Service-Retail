Each test below states its own null/alternative hypothesis, statistic and business interpretation - see `src/statistical_analysis.py`.

```python
import sys; sys.path.insert(0, '../src')
from data_preprocessing import run as preprocess
import statistical_analysis as sa

df, _ = preprocess('../data/coffee_shop_sales.csv')
results = sa.run(df)
results
```

**Interpretation**

1. **Weekday vs weekend t-test** - p = 0.74, not significant. Average spend *per transaction* does not differ by weekend/weekday; any weekend revenue difference would come from transaction *volume*, not basket size.
2. **Store ANOVA** - p < 0.001, significant. Lower Manhattan has a meaningfully higher average transaction value than Astoria; worth a targeted look at Astoria's menu mix or price points.
3. **Chi-square (category x store)** - p < 0.001, significant. Product mix genuinely differs by neighborhood, which supports store-specific (not one-size-fits-all) assortment and promo planning.
4. **Price-quantity correlation** - weak negative correlation (r ≈ -0.12) that is statistically significant only because n is large (~149K rows); the *practical* effect size is small, a reminder that statistical and business significance are not the same thing.
