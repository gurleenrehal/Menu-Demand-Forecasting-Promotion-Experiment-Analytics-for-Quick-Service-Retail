**Reading the experiment design output**

- `power_analysis` says a promo experiment aiming to detect a *5% lift in average transaction value* needs about **5,110 transactions per arm** at 80% power / 5% significance - useful for telling a stakeholder how long a real test would need to run before trusting the result.
- `simulated_ab_test` is a synthetic experiment (uplift is injected, not observed) that exercises the same t-test / confidence-interval workflow a real promo readout would use. It is run at 4,000/arm - below the required 5,110 - which is a deliberate illustration of why under-powering a test is risky even when it happens to come back significant.

**Business recommendations (from the real analysis above)**
1. Treat Coffee and Tea as the demand-forecasting priority - they drive ~66% of revenue, so a 1-point MAPE improvement there is worth more than the same improvement on Branded or Packaged Chocolate.
2. Use recent rolling demand (7-14 day average), not the calendar, as the main forecasting signal - that is what the model actually leans on.
3. Before running a real promo, size the test with the power analysis above rather than picking an arbitrary sample/duration.
4. Investigate why Lower Manhattan's average basket is higher (menu mix vs. pricing vs. customer type) before copying its promo strategy to Astoria - the ANOVA says the stores are not interchangeable.
